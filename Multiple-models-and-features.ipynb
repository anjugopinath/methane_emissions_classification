```python
# ============================================================
# METHANE EMISSION CLASSIFICATION
# Fugitive vs Vented
#
# Models:
#   1. Majority Dummy
#   2. Logistic Regression
#   3. Decision Tree
#   4. Random Forest
#   5. XGBoost
#
# Evaluation:
#   5-Fold Chronological Cross-Validation
#
# Features:
#   Equipment
#   Methane Emission (kg/h)
#   Duration (seconds)
#   Height (m)
#
# Target:
#   Emission Category
#       Vented   = 0
#       Fugitive = 1
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import TimeSeriesSplit

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    matthews_corrcoef,
    roc_curve
)

from xgboost import XGBClassifier


# ============================================================
# 1. LOAD DATA
# ============================================================

print("=" * 80)
print("DATASET")
print("=" * 80)

print("Original shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())


# ============================================================
# 2. KEEP ONLY BINARY CLASSES
# ============================================================

df = df[
    df["Emission Category"].isin(
        ["Fugitive", "Vented"]
    )
].copy()


# ============================================================
# 3. PARSE START TIME AND SORT CHRONOLOGICALLY
# ============================================================

df["Start"] = pd.to_datetime(
    df["Start"],
    errors="coerce"
)

df = df.dropna(
    subset=["Start"]
)

df = df.sort_values(
    by="Start"
).reset_index(drop=True)


print("\nTime range:")

print(
    "Earliest:",
    df["Start"].min()
)

print(
    "Latest:  ",
    df["Start"].max()
)


# ============================================================
# 4. CREATE BINARY TARGET
# ============================================================

# Positive class = Fugitive

df["Target"] = df[
    "Emission Category"
].map({
    "Vented": 0,
    "Fugitive": 1
})


print("\nClass counts:")

print(
    df["Emission Category"]
    .value_counts()
)


print("\nClass proportions:")

print(
    df["Emission Category"]
    .value_counts(normalize=True)
    .round(4)
)


# ============================================================
# 5. DEFINE FEATURES
# ============================================================

categorical_features = [
    "Equipment"
]

numerical_features = [
    "Methane Emission (kg/h)",
    "Duration (seconds)",
    "Height (m)"
]

feature_columns = (
    categorical_features
    + numerical_features
)


# ============================================================
# 6. CLEAN FEATURES
# ============================================================

for col in numerical_features:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


df = df.dropna(
    subset=feature_columns + ["Target"]
).reset_index(drop=True)


X = df[feature_columns]

y = df["Target"].astype(int)


print("\nFinal number of samples:", len(df))

print("\nFeatures used:")

for feature in feature_columns:
    print(" -", feature)


# ============================================================
# 7. PREPROCESSING
# ============================================================

# One-hot encode Equipment
# Scale numerical variables
#
# Scaling is important for Logistic Regression.
# It has little effect on tree-based models.

preprocessor = ColumnTransformer(
    transformers=[

        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),

        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)


# ============================================================
# 8. DEFINE MODELS
# ============================================================

models = {

    # --------------------------------------------------------
    # Majority baseline
    # --------------------------------------------------------

    "Majority Dummy": DummyClassifier(
        strategy="most_frequent"
    ),


    # --------------------------------------------------------
    # Logistic Regression
    # --------------------------------------------------------

    "Logistic Regression": LogisticRegression(
        class_weight="balanced",
        max_iter=5000,
        random_state=42
    ),


    # --------------------------------------------------------
    # Decision Tree
    # --------------------------------------------------------

    "Decision Tree": DecisionTreeClassifier(
        class_weight="balanced",
        min_samples_leaf=2,
        random_state=42
    ),


    # --------------------------------------------------------
    # Random Forest
    # --------------------------------------------------------

    "Random Forest": RandomForestClassifier(
        n_estimators=500,
        class_weight="balanced",
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ),


    # --------------------------------------------------------
    # XGBoost
    # --------------------------------------------------------

    "XGBoost": XGBClassifier(

        n_estimators=500,

        learning_rate=0.05,

        max_depth=5,

        min_child_weight=2,

        subsample=0.8,

        colsample_bytree=0.8,

        objective="binary:logistic",

        eval_metric="logloss",

        random_state=42,

        n_jobs=-1
    )
}


# ============================================================
# 9. DEFINE 5-FOLD CHRONOLOGICAL CV
# ============================================================

tscv = TimeSeriesSplit(
    n_splits=5
)


# ============================================================
# 10. STORAGE
# ============================================================

all_fold_results = []

model_predictions = {}

roc_cv_results = {}


# ============================================================
# 11. RUN ALL MODELS
# ============================================================

for model_name, classifier in models.items():

    print("\n")
    print("=" * 80)
    print("MODEL:", model_name)
    print("=" * 80)


    # --------------------------------------------------------
    # Store pooled predictions
    # --------------------------------------------------------

    all_true = []
    all_pred = []
    all_prob = []


    # --------------------------------------------------------
    # Store fold ROC curves
    # --------------------------------------------------------

    fold_roc_curves = []

    fold_auc_scores = []


    # ========================================================
    # 5 CHRONOLOGICAL FOLDS
    # ========================================================

    for fold, (train_idx, test_idx) in enumerate(
        tscv.split(X),
        start=1
    ):

        X_train = X.iloc[
            train_idx
        ]

        X_test = X.iloc[
            test_idx
        ]

        y_train = y.iloc[
            train_idx
        ]

        y_test = y.iloc[
            test_idx
        ]


        # ----------------------------------------------------
        # Temporal boundaries
        # ----------------------------------------------------

        train_start = df.iloc[
            train_idx
        ]["Start"].min()

        train_end = df.iloc[
            train_idx
        ]["Start"].max()

        test_start = df.iloc[
            test_idx
        ]["Start"].min()

        test_end = df.iloc[
            test_idx
        ]["Start"].max()


        print("\nFold", fold)

        print(
            "Train:",
            train_start,
            "-->",
            train_end
        )

        print(
            "Test :",
            test_start,
            "-->",
            test_end
        )

        print(
            f"Train size = {len(train_idx)}, "
            f"Test size = {len(test_idx)}"
        )


        # ----------------------------------------------------
        # Ensure training contains both classes
        # ----------------------------------------------------

        if y_train.nunique() < 2:

            print(
                "Skipping fold because training "
                "contains only one class."
            )

            continue


        # ----------------------------------------------------
        # Create NEW pipeline for every fold
        #
        # This prevents preprocessing leakage.
        # ----------------------------------------------------

        pipeline = Pipeline([

            (
                "preprocessor",
                clone(preprocessor)
            ),

            (
                "classifier",
                clone(classifier)
            )
        ])


        # ----------------------------------------------------
        # Train
        # ----------------------------------------------------

        pipeline.fit(
            X_train,
            y_train
        )


        # ----------------------------------------------------
        # Predictions
        # ----------------------------------------------------

        y_pred = pipeline.predict(
            X_test
        )


        # ----------------------------------------------------
        # Probability for Fugitive = class 1
        # ----------------------------------------------------

        probability_matrix = (
            pipeline.predict_proba(
                X_test
            )
        )


        classes = (
            pipeline
            .named_steps["classifier"]
            .classes_
        )


        if 1 in classes:

            positive_class_index = (
                np.where(
                    classes == 1
                )[0][0]
            )

            y_prob = probability_matrix[
                :,
                positive_class_index
            ]

        else:

            # Defensive fallback
            y_prob = np.zeros(
                len(y_test)
            )


        # ====================================================
        # METRICS
        # ====================================================

        accuracy = accuracy_score(
            y_test,
            y_pred
        )


        balanced_accuracy = (
            balanced_accuracy_score(
                y_test,
                y_pred
            )
        )


        precision = precision_score(
            y_test,
            y_pred,
            zero_division=0
        )


        recall = recall_score(
            y_test,
            y_pred,
            zero_division=0
        )


        f1 = f1_score(
            y_test,
            y_pred,
            zero_division=0
        )


        mcc = matthews_corrcoef(
            y_test,
            y_pred
        )


        # ----------------------------------------------------
        # Confusion matrix
        # ----------------------------------------------------

        cm = confusion_matrix(
            y_test,
            y_pred,
            labels=[0, 1]
        )


        tn, fp, fn, tp = cm.ravel()


        # ----------------------------------------------------
        # Specificity
        # ----------------------------------------------------

        if (tn + fp) > 0:

            specificity = (
                tn / (tn + fp)
            )

        else:

            specificity = np.nan


        # ----------------------------------------------------
        # AUROC and AUPRC
        # ----------------------------------------------------

        if y_test.nunique() == 2:

            roc_auc = roc_auc_score(
                y_test,
                y_prob
            )


            pr_auc = (
                average_precision_score(
                    y_test,
                    y_prob
                )
            )


            # ROC curve for this fold

            fpr, tpr, _ = roc_curve(
                y_test,
                y_prob
            )


            fold_roc_curves.append(
                (fpr, tpr)
            )


            fold_auc_scores.append(
                roc_auc
            )


        else:

            roc_auc = np.nan

            pr_auc = np.nan


        # ----------------------------------------------------
        # Save fold results
        # ----------------------------------------------------

        all_fold_results.append({

            "Model": model_name,

            "Fold": fold,

            "Accuracy": accuracy,

            "Balanced Accuracy":
                balanced_accuracy,

            "Precision": precision,

            "Recall": recall,

            "Specificity": specificity,

            "F1": f1,

            "AUROC": roc_auc,

            "AUPRC": pr_auc,

            "MCC": mcc
        })


        # ----------------------------------------------------
        # Store out-of-time predictions
        # ----------------------------------------------------

        all_true.extend(
            y_test.to_numpy()
        )

        all_pred.extend(
            y_pred
        )

        all_prob.extend(
            y_prob
        )


    # ========================================================
    # STORE RESULTS FOR MODEL
    # ========================================================

    model_predictions[
        model_name
    ] = {

        "y_true": np.array(
            all_true
        ),

        "y_pred": np.array(
            all_pred
        ),

        "y_prob": np.array(
            all_prob
        )
    }


    roc_cv_results[
        model_name
    ] = {

        "roc_curves":
            fold_roc_curves,

        "auc_scores":
            np.array(
                fold_auc_scores
            )
    }


# ============================================================
# 12. FOLD-WISE RESULTS
# ============================================================

fold_results_df = pd.DataFrame(
    all_fold_results
)


print("\n")
print("=" * 120)
print("FOLD-WISE RESULTS")
print("=" * 120)


print(
    fold_results_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 13. 5-FOLD MEAN ± STANDARD DEVIATION
# ============================================================

metric_columns = [

    "Accuracy",

    "Balanced Accuracy",

    "Precision",

    "Recall",

    "Specificity",

    "F1",

    "AUROC",

    "AUPRC",

    "MCC"
]


summary_mean = (

    fold_results_df

    .groupby("Model")[
        metric_columns
    ]

    .mean()
)


summary_std = (

    fold_results_df

    .groupby("Model")[
        metric_columns
    ]

    .std()
)


# ============================================================
# 14. FORMAT MEAN ± SD TABLE
# ============================================================

summary_formatted = pd.DataFrame(
    index=summary_mean.index
)


for metric in metric_columns:

    summary_formatted[
        metric
    ] = [

        (
            f"{summary_mean.loc[model, metric]:.3f}"
            f" ± "
            f"{summary_std.loc[model, metric]:.3f}"
        )

        for model in summary_mean.index
    ]


# Reorder models

model_order = [

    "Majority Dummy",

    "Logistic Regression",

    "Decision Tree",

    "Random Forest",

    "XGBoost"
]


summary_formatted = (
    summary_formatted
    .reindex(model_order)
)


print("\n")
print("=" * 120)

print(
    "5-FOLD CHRONOLOGICAL CV: "
    "MEAN ± STANDARD DEVIATION"
)

print("=" * 120)


print(
    summary_formatted.to_string()
)


# ============================================================
# 15. POOLED OUT-OF-TIME METRICS
# ============================================================

pooled_results = []


for model_name in model_order:

    predictions = model_predictions[
        model_name
    ]

    y_true = predictions[
        "y_true"
    ]

    y_pred = predictions[
        "y_pred"
    ]

    y_prob = predictions[
        "y_prob"
    ]


    # --------------------------------------------------------
    # Basic metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_true,
        y_pred
    )


    balanced_accuracy = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )


    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )


    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )


    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )


    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )


    # --------------------------------------------------------
    # Specificity
    # --------------------------------------------------------

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )


    tn, fp, fn, tp = cm.ravel()


    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else np.nan
    )


    # --------------------------------------------------------
    # AUROC / AUPRC
    # --------------------------------------------------------

    if len(
        np.unique(y_true)
    ) == 2:

        roc_auc = roc_auc_score(
            y_true,
            y_prob
        )

        pr_auc = (
            average_precision_score(
                y_true,
                y_prob
            )
        )

    else:

        roc_auc = np.nan

        pr_auc = np.nan


    pooled_results.append({

        "Model":
            model_name,

        "Accuracy":
            accuracy,

        "Balanced Accuracy":
            balanced_accuracy,

        "Precision":
            precision,

        "Recall":
            recall,

        "Specificity":
            specificity,

        "F1":
            f1,

        "AUROC":
            roc_auc,

        "AUPRC":
            pr_auc,

        "MCC":
            mcc
    })


pooled_df = pd.DataFrame(
    pooled_results
)


print("\n")
print("=" * 120)

print(
    "POOLED OUT-OF-TIME PERFORMANCE"
)

print("=" * 120)


print(
    pooled_df
    .round(4)
    .to_string(
        index=False
    )
)


# ============================================================
# 16. SAVE TABLES
# ============================================================

fold_results_df.to_csv(
    "methane_classifier_fold_results.csv",
    index=False
)


summary_formatted.to_csv(
    "methane_classifier_5fold_mean_std.csv"
)


pooled_df.to_csv(
    "methane_classifier_pooled_results.csv",
    index=False
)


# ============================================================
# 17. ROC PLOT 1:
#     POOLED OUT-OF-TIME ROC
# ============================================================

plt.figure(
    figsize=(8, 7)
)


for model_name in model_order:

    # Skip dummy because chance line is already shown
    if model_name == "Majority Dummy":
        continue


    predictions = model_predictions[
        model_name
    ]


    y_true = predictions[
        "y_true"
    ]

    y_prob = predictions[
        "y_prob"
    ]


    if len(
        np.unique(y_true)
    ) < 2:

        continue


    fpr, tpr, _ = roc_curve(
        y_true,
        y_prob
    )


    auc_value = roc_auc_score(
        y_true,
        y_prob
    )


    plt.plot(
        fpr,
        tpr,
        linewidth=2,
        label=(
            f"{model_name} "
            f"(AUC = {auc_value:.3f})"
        )
    )


# Chance line

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Chance (AUC = 0.500)"
)


plt.xlabel(
    "False Positive Rate",
    fontsize=13
)


plt.ylabel(
    "True Positive Rate",
    fontsize=13
)


plt.title(
    "Pooled Out-of-Time ROC Curves",
    fontsize=14
)


plt.xlim(
    0,
    1
)


plt.ylim(
    0,
    1.02
)


plt.legend(
    loc="lower right",
    fontsize=10
)


plt.grid(
    alpha=0.2
)


plt.tight_layout()


plt.savefig(
    "roc_pooled_out_of_time.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 18. ROC PLOT 2:
#     5-FOLD MEAN ROC ± STANDARD DEVIATION
# ============================================================

plt.figure(
    figsize=(8, 7)
)


# Every fold has different FPR positions.
# Interpolate all ROC curves onto a common grid.

mean_fpr = np.linspace(
    0,
    1,
    200
)


for model_name in model_order:

    # Skip dummy from ROC figure
    if model_name == "Majority Dummy":
        continue


    roc_data = roc_cv_results[
        model_name
    ]


    roc_curves = roc_data[
        "roc_curves"
    ]


    auc_scores = roc_data[
        "auc_scores"
    ]


    if len(
        roc_curves
    ) == 0:

        continue


    interpolated_tprs = []


    # --------------------------------------------------------
    # Interpolate ROC of each fold
    # --------------------------------------------------------

    for fpr, tpr in roc_curves:

        interp_tpr = np.interp(
            mean_fpr,
            fpr,
            tpr
        )

        interp_tpr[0] = 0.0

        interpolated_tprs.append(
            interp_tpr
        )


    interpolated_tprs = np.array(
        interpolated_tprs
    )


    # --------------------------------------------------------
    # Mean TPR
    # --------------------------------------------------------

    mean_tpr = np.mean(
        interpolated_tprs,
        axis=0
    )

    mean_tpr[-1] = 1.0


    # --------------------------------------------------------
    # SD of TPR
    # --------------------------------------------------------

    std_tpr = np.std(
        interpolated_tprs,
        axis=0,
        ddof=1
    )


    lower_tpr = np.maximum(
        mean_tpr - std_tpr,
        0
    )


    upper_tpr = np.minimum(
        mean_tpr + std_tpr,
        1
    )


    # --------------------------------------------------------
    # Mean ± SD AUROC
    # --------------------------------------------------------

    mean_auc = np.mean(
        auc_scores
    )


    std_auc = np.std(
        auc_scores,
        ddof=1
    )


    # --------------------------------------------------------
    # Plot mean curve
    # --------------------------------------------------------

    line, = plt.plot(

        mean_fpr,

        mean_tpr,

        linewidth=2,

        label=(
            f"{model_name} "
            f"(AUC = "
            f"{mean_auc:.3f} ± "
            f"{std_auc:.3f})"
        )
    )


    # --------------------------------------------------------
    # ± 1 SD band
    # --------------------------------------------------------

    plt.fill_between(

        mean_fpr,

        lower_tpr,

        upper_tpr,

        alpha=0.12,

        color=line.get_color()
    )


# Chance line

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Chance (AUC = 0.500)"
)


plt.xlabel(
    "False Positive Rate",
    fontsize=13
)


plt.ylabel(
    "True Positive Rate",
    fontsize=13
)


plt.title(
    "Mean ROC Curves – "
    "5-Fold Chronological CV",
    fontsize=14
)


plt.xlim(
    0,
    1
)


plt.ylim(
    0,
    1.02
)


plt.legend(
    loc="lower right",
    fontsize=10
)


plt.grid(
    alpha=0.2
)


plt.tight_layout()


plt.savefig(
    "roc_5fold_mean_std.png",
    dpi=300,
    bbox_inches="tight"
)


plt.show()


# ============================================================
# 19. PRINT INDIVIDUAL FOLD AUROC VALUES
# ============================================================

print("\n")
print("=" * 80)
print("AUROC FOR EACH CHRONOLOGICAL FOLD")
print("=" * 80)


for model_name in model_order:

    auc_scores = roc_cv_results[
        model_name
    ]["auc_scores"]


    if len(auc_scores) == 0:
        continue


    print(
        f"\n{model_name}:"
    )


    for i, auc_value in enumerate(
        auc_scores,
        start=1
    ):

        print(
            f"  Fold {i}: "
            f"{auc_value:.4f}"
        )


    print(
        "  Mean ± SD: "
        f"{np.mean(auc_scores):.4f}"
        " ± "
        f"{np.std(auc_scores, ddof=1):.4f}"
    )


print("\nDone.")
```

```python
# ============================================================
# METHANE EMISSION BINARY CLASSIFICATION
# Fugitive vs Vented
#
# Rolling chronological evaluation:
#   Train = previous 30 calendar days
#   Test  = following 7 calendar days
#   Step  = 7 days
#   Use ALL complete windows available
#
# Models:
#   1. Majority Dummy
#   2. Logistic Regression
#   3. Decision Tree
#   4. Random Forest
#   5. XGBoost
#
# Features:
#   Equipment
#   Methane Emission (kg/h)
#   Duration (seconds)
#   Height (m)
#
# Target:
#   Emission Category
#       Fugitive = 1
#       Vented   = 0
#
# Start is used ONLY for chronological splitting.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    matthews_corrcoef,
    roc_curve
)

from xgboost import XGBClassifier


# ============================================================
# 1. SETTINGS
# ============================================================

#FILE_PATH = "SEALS-4_Month_Data_Summary.xlsx"

TRAIN_DAYS = 30
TEST_DAYS = 7
STEP_DAYS = 7

RANDOM_STATE = 42


# ============================================================
# 2. LOAD DATA
# ============================================================

#df = pd.read_excel(FILE_PATH)

print("=" * 90)
print("DATASET INFORMATION")
print("=" * 90)

print("Original shape:", df.shape)

print("\nColumns:")
for col in df.columns:
    print(" -", col)


# ============================================================
# 3. KEEP ONLY FUGITIVE AND VENTED
# ============================================================

df = df[
    df["Emission Category"].isin(
        ["Fugitive", "Vented"]
    )
].copy()


# ============================================================
# 4. PARSE START TIME
# ============================================================

df["Start"] = pd.to_datetime(
    df["Start"],
    errors="coerce"
)

df = df.dropna(
    subset=["Start"]
)


# ============================================================
# 5. SORT CHRONOLOGICALLY
# ============================================================

df = (
    df
    .sort_values("Start")
    .reset_index(drop=True)
)


print("\nTime span:")
print("First event:", df["Start"].min())
print("Last event :", df["Start"].max())


# ============================================================
# 6. CREATE BINARY TARGET
# ============================================================

# Positive class = Fugitive

df["Target"] = (
    df["Emission Category"]
    .map({
        "Vented": 0,
        "Fugitive": 1
    })
)


print("\nClass distribution:")
print(
    df["Emission Category"]
    .value_counts()
)

print("\nClass proportions:")
print(
    df["Emission Category"]
    .value_counts(normalize=True)
    .round(4)
)


# ============================================================
# 7. DEFINE FEATURES
# ============================================================

categorical_features = [
    "Equipment"
]

numerical_features = [
    "Methane Emission (kg/h)",
    "Duration (seconds)",
    "Height (m)"
]

feature_columns = (
    categorical_features
    + numerical_features
)


# ============================================================
# 8. CLEAN FEATURES
# ============================================================

for col in numerical_features:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )


df = df.dropna(
    subset=feature_columns + ["Target"]
).reset_index(drop=True)


X = df[feature_columns].copy()

y = df["Target"].astype(int)


print("\nFinal binary dataset size:", len(df))

print("\nFeatures used:")
for feature in feature_columns:
    print(" -", feature)


# ============================================================
# 9. PREPROCESSING
# ============================================================

# Equipment:
#   One-hot encoded
#
# Numerical variables:
#   Standardized
#
# Scaling mainly helps Logistic Regression.
# Tree models are largely insensitive to scaling.

preprocessor = ColumnTransformer(

    transformers=[

        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),

        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ]
)


# ============================================================
# 10. DEFINE CLASSIFIERS
# ============================================================

models = {

    "Majority Dummy":
        DummyClassifier(
            strategy="most_frequent"
        ),

    "Logistic Regression":
        LogisticRegression(
            class_weight="balanced",
            max_iter=5000,
            random_state=RANDOM_STATE
        ),

    "Decision Tree":
        DecisionTreeClassifier(
            class_weight="balanced",
            min_samples_leaf=2,
            random_state=RANDOM_STATE
        ),

    "Random Forest":
        RandomForestClassifier(
            n_estimators=500,
            class_weight="balanced",
            min_samples_split=5,
            min_samples_leaf=2,
            random_state=RANDOM_STATE,
            n_jobs=-1
        ),

    "XGBoost":
        XGBClassifier(
            n_estimators=500,
            learning_rate=0.05,
            max_depth=5,
            min_child_weight=2,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="binary:logistic",
            eval_metric="logloss",
            tree_method="hist",
            random_state=RANDOM_STATE,
            n_jobs=-1,
            verbosity=0
        )
}


model_order = [
    "Majority Dummy",
    "Logistic Regression",
    "Decision Tree",
    "Random Forest",
    "XGBoost"
]


# ============================================================
# 11. CREATE TIME-BASED ROLLING SPLITS
# ============================================================

def create_time_splits(
    dataframe,
    time_column,
    train_days=30,
    test_days=7,
    step_days=7
):

    """
    Create all possible complete rolling chronological windows.

    Example:

    Fold 1:
        Train: Day 0  -> Day 30
        Test : Day 30 -> Day 37

    Fold 2:
        Train: Day 7  -> Day 37
        Test : Day 37 -> Day 44

    Fold 3:
        Train: Day 14 -> Day 44
        Test : Day 44 -> Day 51

    etc.

    Train and test intervals are left-inclusive
    and right-exclusive:

        train_start <= time < train_end
        test_start  <= time < test_end
    """

    splits = []

    # Use calendar-day boundaries
    first_day = (
        dataframe[time_column]
        .min()
        .normalize()
    )

    last_day = (
        dataframe[time_column]
        .max()
        .normalize()
    )

    # Exclusive end of dataset's final calendar day
    data_end = (
        last_day
        + pd.Timedelta(days=1)
    )


    # First test begins after initial 30-day training period
    test_start = (
        first_day
        + pd.Timedelta(days=train_days)
    )


    fold = 1


    while True:

        train_end = test_start

        train_start = (
            train_end
            - pd.Timedelta(days=train_days)
        )

        test_end = (
            test_start
            + pd.Timedelta(days=test_days)
        )


        # Stop if there is not a complete test window
        if test_end > data_end:
            break


        train_mask = (

            (dataframe[time_column] >= train_start)

            &

            (dataframe[time_column] < train_end)
        )


        test_mask = (

            (dataframe[time_column] >= test_start)

            &

            (dataframe[time_column] < test_end)
        )


        train_idx = dataframe.index[
            train_mask
        ].to_numpy()


        test_idx = dataframe.index[
            test_mask
        ].to_numpy()


        # Only keep windows containing observations
        if (
            len(train_idx) > 0
            and
            len(test_idx) > 0
        ):

            splits.append({

                "Fold": fold,

                "Train Start": train_start,

                "Train End": train_end,

                "Test Start": test_start,

                "Test End": test_end,

                "Train Index": train_idx,

                "Test Index": test_idx
            })


            fold += 1


        # Move forward by one test period
        test_start = (
            test_start
            + pd.Timedelta(days=step_days)
        )


    return splits


# ============================================================
# 12. GENERATE ALL AVAILABLE ROLLING WINDOWS
# ============================================================

time_splits = create_time_splits(

    dataframe=df,

    time_column="Start",

    train_days=TRAIN_DAYS,

    test_days=TEST_DAYS,

    step_days=STEP_DAYS
)


print("\n")
print("=" * 110)

print(
    f"{TRAIN_DAYS}-DAY TRAIN + "
    f"{TEST_DAYS}-DAY TEST "
    "ROLLING WINDOWS"
)

print("=" * 110)


print(
    "\nNumber of available folds:",
    len(time_splits)
)


for split in time_splits:

    print(
        f"\nFold {split['Fold']}"
    )

    print(
        "Train:",
        split["Train Start"],
        "-->",
        split["Train End"]
    )

    print(
        "Test :",
        split["Test Start"],
        "-->",
        split["Test End"]
    )

    print(
        "Train samples:",
        len(split["Train Index"])
    )

    print(
        "Test samples :",
        len(split["Test Index"])
    )


if len(time_splits) == 0:

    raise ValueError(
        "No valid rolling windows were created. "
        "Check the timestamp range."
    )


# ============================================================
# 13. STORAGE
# ============================================================

all_fold_results = []

model_predictions = {}

roc_cv_results = {}


# ============================================================
# 14. RUN ALL MODELS
# ============================================================

for model_name, classifier in models.items():

    print("\n")
    print("=" * 90)
    print("MODEL:", model_name)
    print("=" * 90)


    # Pooled predictions across all test windows
    all_true = []
    all_pred = []
    all_prob = []


    # Fold-wise ROC information
    fold_roc_curves = []
    fold_auc_scores = []
    fold_numbers_for_auc = []


    # ========================================================
    # LOOP OVER ROLLING WINDOWS
    # ========================================================

    for split in time_splits:

        fold = split["Fold"]

        train_idx = split["Train Index"]
        test_idx = split["Test Index"]


        X_train = X.loc[train_idx]
        X_test = X.loc[test_idx]

        y_train = y.loc[train_idx]
        y_test = y.loc[test_idx]


        print("\nFold", fold)

        print(
            "Training:",
            split["Train Start"],
            "-->",
            split["Train End"]
        )

        print(
            "Testing :",
            split["Test Start"],
            "-->",
            split["Test End"]
        )

        print(
            "Train samples:",
            len(y_train)
        )

        print(
            "Test samples :",
            len(y_test)
        )

        print(
            "Train classes:",
            y_train.value_counts()
            .sort_index()
            .to_dict()
        )

        print(
            "Test classes:",
            y_test.value_counts()
            .sort_index()
            .to_dict()
        )


        # ----------------------------------------------------
        # Training must contain both classes
        # ----------------------------------------------------

        if y_train.nunique() < 2:

            print(
                "WARNING: skipped because training "
                "window contains only one class."
            )

            continue


        # ----------------------------------------------------
        # Fresh model + preprocessing for every fold
        # ----------------------------------------------------

        pipeline = Pipeline([

            (
                "preprocessor",
                clone(preprocessor)
            ),

            (
                "classifier",
                clone(classifier)
            )
        ])


        # ----------------------------------------------------
        # TRAIN
        # ----------------------------------------------------

        pipeline.fit(
            X_train,
            y_train
        )


        # ----------------------------------------------------
        # PREDICT
        # ----------------------------------------------------

        y_pred = pipeline.predict(
            X_test
        )


        probability_matrix = (
            pipeline.predict_proba(
                X_test
            )
        )


        classes = (
            pipeline
            .named_steps["classifier"]
            .classes_
        )


        positive_index = np.where(
            classes == 1
        )[0][0]


        y_prob = probability_matrix[
            :,
            positive_index
        ]


        # ====================================================
        # METRICS
        # ====================================================

        accuracy = accuracy_score(
            y_test,
            y_pred
        )


        balanced_accuracy = (
            balanced_accuracy_score(
                y_test,
                y_pred
            )
        )


        precision = precision_score(
            y_test,
            y_pred,
            pos_label=1,
            zero_division=0
        )


        recall = recall_score(
            y_test,
            y_pred,
            pos_label=1,
            zero_division=0
        )


        f1 = f1_score(
            y_test,
            y_pred,
            pos_label=1,
            zero_division=0
        )


        f1_macro = f1_score(
            y_test,
            y_pred,
            average="macro",
            zero_division=0
        )


        mcc = matthews_corrcoef(
            y_test,
            y_pred
        )


        # ====================================================
        # CONFUSION MATRIX
        # ====================================================

        cm = confusion_matrix(
            y_test,
            y_pred,
            labels=[0, 1]
        )


        tn, fp, fn, tp = cm.ravel()


        specificity = (

            tn / (tn + fp)

            if (tn + fp) > 0

            else np.nan
        )


        # ====================================================
        # AUROC / AUPRC
        # ====================================================

        if y_test.nunique() == 2:

            roc_auc = roc_auc_score(
                y_test,
                y_prob
            )


            pr_auc = average_precision_score(
                y_test,
                y_prob
            )


            fpr, tpr, _ = roc_curve(
                y_test,
                y_prob
            )


            fold_roc_curves.append(
                (fpr, tpr)
            )


            fold_auc_scores.append(
                roc_auc
            )


            fold_numbers_for_auc.append(
                fold
            )


        else:

            roc_auc = np.nan

            pr_auc = np.nan


            print(
                "WARNING: Test window contains "
                "only one class. "
                "AUROC/AUPRC not calculated."
            )


        # ====================================================
        # STORE FOLD METRICS
        # ====================================================

        all_fold_results.append({

            "Model":
                model_name,

            "Fold":
                fold,

            "Train Start":
                split["Train Start"],

            "Train End":
                split["Train End"],

            "Test Start":
                split["Test Start"],

            "Test End":
                split["Test End"],

            "Train Samples":
                len(y_train),

            "Test Samples":
                len(y_test),

            "Accuracy":
                accuracy,

            "Balanced Accuracy":
                balanced_accuracy,

            "Precision":
                precision,

            "Recall":
                recall,

            "Specificity":
                specificity,

            "F1":
                f1,

            "F1 Macro":
                f1_macro,

            "AUROC":
                roc_auc,

            "AUPRC":
                pr_auc,

            "MCC":
                mcc
        })


        # ====================================================
        # STORE OUT-OF-TIME PREDICTIONS
        # ====================================================

        all_true.extend(
            y_test.to_numpy()
        )


        all_pred.extend(
            y_pred
        )


        all_prob.extend(
            y_prob
        )


    # ========================================================
    # STORE RESULTS FOR THIS MODEL
    # ========================================================

    model_predictions[
        model_name
    ] = {

        "y_true":
            np.array(all_true),

        "y_pred":
            np.array(all_pred),

        "y_prob":
            np.array(all_prob)
    }


    roc_cv_results[
        model_name
    ] = {

        "roc_curves":
            fold_roc_curves,

        "auc_scores":
            np.array(
                fold_auc_scores
            ),

        "fold_numbers":
            fold_numbers_for_auc
    }


# ============================================================
# 15. FOLD-WISE RESULTS TABLE
# ============================================================

fold_results_df = pd.DataFrame(
    all_fold_results
)


display_columns = [

    "Model",
    "Fold",

    "Train Samples",
    "Test Samples",

    "Accuracy",
    "Balanced Accuracy",

    "Precision",
    "Recall",
    "Specificity",

    "F1",
    "F1 Macro",

    "AUROC",
    "AUPRC",

    "MCC"
]


print("\n")
print("=" * 140)
print("FOLD-WISE ROLLING-WINDOW RESULTS")
print("=" * 140)


print(

    fold_results_df[
        display_columns
    ]

    .round(4)

    .to_string(
        index=False
    )
)


# ============================================================
# 16. MEAN ± STANDARD DEVIATION TABLE
# ============================================================

metric_columns = [

    "Accuracy",

    "Balanced Accuracy",

    "Precision",

    "Recall",

    "Specificity",

    "F1",

    "F1 Macro",

    "AUROC",

    "AUPRC",

    "MCC"
]


summary_mean = (

    fold_results_df

    .groupby(
        "Model"
    )[metric_columns]

    .mean()
)


summary_std = (

    fold_results_df

    .groupby(
        "Model"
    )[metric_columns]

    .std()
)


summary_formatted = pd.DataFrame(
    index=summary_mean.index
)


for metric in metric_columns:

    summary_formatted[
        metric
    ] = [

        (
            f"{summary_mean.loc[model, metric]:.3f}"
            f" ± "
            f"{summary_std.loc[model, metric]:.3f}"
        )

        for model in summary_mean.index
    ]


summary_formatted = (
    summary_formatted
    .reindex(model_order)
)


print("\n")
print("=" * 140)

print(
    f"{TRAIN_DAYS}-DAY TRAIN + "
    f"{TEST_DAYS}-DAY TEST "
    "ROLLING CV: MEAN ± STANDARD DEVIATION"
)

print("=" * 140)


print(
    summary_formatted.to_string()
)


# ============================================================
# 17. NUMERIC SUMMARY TABLE
#     Useful for plotting / later statistical analysis
# ============================================================

summary_numeric = pd.DataFrame()


for model in model_order:

    if model not in summary_mean.index:
        continue


    row = {
        "Model": model
    }


    for metric in metric_columns:

        row[
            metric + " Mean"
        ] = summary_mean.loc[
            model,
            metric
        ]


        row[
            metric + " SD"
        ] = summary_std.loc[
            model,
            metric
        ]


    summary_numeric = pd.concat(

        [
            summary_numeric,
            pd.DataFrame([row])
        ],

        ignore_index=True
    )


# ============================================================
# 18. POOLED OUT-OF-TIME PERFORMANCE
# ============================================================

pooled_results = []


for model_name in model_order:

    predictions = model_predictions[
        model_name
    ]


    y_true = predictions[
        "y_true"
    ]

    y_pred = predictions[
        "y_pred"
    ]

    y_prob = predictions[
        "y_prob"
    ]


    if len(y_true) == 0:
        continue


    accuracy = accuracy_score(
        y_true,
        y_pred
    )


    balanced_accuracy = (
        balanced_accuracy_score(
            y_true,
            y_pred
        )
    )


    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )


    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )


    specificity_cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )


    tn, fp, fn, tp = (
        specificity_cm.ravel()
    )


    specificity = (

        tn / (tn + fp)

        if (tn + fp) > 0

        else np.nan
    )


    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )


    f1_macro = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )


    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )


    if len(
        np.unique(y_true)
    ) == 2:

        roc_auc = roc_auc_score(
            y_true,
            y_prob
        )


        pr_auc = average_precision_score(
            y_true,
            y_prob
        )


    else:

        roc_auc = np.nan
        pr_auc = np.nan


    pooled_results.append({

        "Model":
            model_name,

        "Accuracy":
            accuracy,

        "Balanced Accuracy":
            balanced_accuracy,

        "Precision":
            precision,

        "Recall":
            recall,

        "Specificity":
            specificity,

        "F1":
            f1,

        "F1 Macro":
            f1_macro,

        "AUROC":
            roc_auc,

        "AUPRC":
            pr_auc,

        "MCC":
            mcc
    })


pooled_df = pd.DataFrame(
    pooled_results
)


print("\n")
print("=" * 140)
print("POOLED OUT-OF-TIME PERFORMANCE")
print("=" * 140)


print(

    pooled_df

    .round(4)

    .to_string(
        index=False
    )
)


# ============================================================
# 19. SAVE TABLES
# ============================================================

fold_results_df.to_csv(

    "methane_rolling_fold_results.csv",

    index=False
)


summary_formatted.to_csv(

    "methane_rolling_mean_std.csv"
)


summary_numeric.to_csv(

    "methane_rolling_mean_std_numeric.csv",

    index=False
)


pooled_df.to_csv(

    "methane_rolling_pooled_results.csv",

    index=False
)


# ============================================================
# 20. ROC FIGURE 1:
#     POOLED OUT-OF-TIME ROC
# ============================================================

plt.figure(
    figsize=(8, 7)
)


for model_name in model_order:

    # Dummy is represented by the chance line
    if model_name == "Majority Dummy":
        continue


    predictions = model_predictions[
        model_name
    ]


    y_true = predictions[
        "y_true"
    ]


    y_prob = predictions[
        "y_prob"
    ]


    if len(
        np.unique(y_true)
    ) < 2:

        continue


    fpr, tpr, _ = roc_curve(
        y_true,
        y_prob
    )


    auc_value = roc_auc_score(
        y_true,
        y_prob
    )


    plt.plot(

        fpr,

        tpr,

        linewidth=2,

        label=(
            f"{model_name} "
            f"(AUC = {auc_value:.3f})"
        )
    )


# Chance line

plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--",

    linewidth=1.5,

    label="Chance (AUC = 0.500)"
)


plt.xlabel(
    "False Positive Rate",
    fontsize=13
)


plt.ylabel(
    "True Positive Rate",
    fontsize=13
)


plt.title(

    "Pooled Out-of-Time ROC\n"
    f"{TRAIN_DAYS}-Day Train + "
    f"{TEST_DAYS}-Day Test",

    fontsize=14
)


plt.xlim(0, 1)

plt.ylim(0, 1.02)


plt.legend(
    loc="lower right",
    fontsize=10
)


plt.grid(
    alpha=0.2
)


plt.tight_layout()


plt.savefig(

    "roc_pooled_out_of_time.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ============================================================
# 21. ROC FIGURE 2:
#     MEAN ROC ± 1 STANDARD DEVIATION
#     ACROSS ALL ROLLING TEST WINDOWS
# ============================================================

plt.figure(
    figsize=(8, 7)
)


# Common FPR grid for averaging curves

mean_fpr = np.linspace(
    0,
    1,
    300
)


for model_name in model_order:

    if model_name == "Majority Dummy":
        continue


    roc_data = roc_cv_results[
        model_name
    ]


    roc_curves = roc_data[
        "roc_curves"
    ]


    auc_scores = roc_data[
        "auc_scores"
    ]


    if len(roc_curves) == 0:
        continue


    interpolated_tprs = []


    # --------------------------------------------------------
    # Interpolate fold ROC curves to common FPR positions
    # --------------------------------------------------------

    for fpr, tpr in roc_curves:

        interp_tpr = np.interp(
            mean_fpr,
            fpr,
            tpr
        )


        interp_tpr[0] = 0.0


        interpolated_tprs.append(
            interp_tpr
        )


    interpolated_tprs = np.array(
        interpolated_tprs
    )


    # --------------------------------------------------------
    # Mean ROC
    # --------------------------------------------------------

    mean_tpr = np.mean(
        interpolated_tprs,
        axis=0
    )


    mean_tpr[0] = 0.0
    mean_tpr[-1] = 1.0


    # --------------------------------------------------------
    # Standard deviation of TPR
    # --------------------------------------------------------

    if len(
        interpolated_tprs
    ) > 1:

        std_tpr = np.std(
            interpolated_tprs,
            axis=0,
            ddof=1
        )

    else:

        std_tpr = np.zeros_like(
            mean_tpr
        )


    lower_tpr = np.maximum(
        mean_tpr - std_tpr,
        0
    )


    upper_tpr = np.minimum(
        mean_tpr + std_tpr,
        1
    )


    # --------------------------------------------------------
    # Fold-wise mean ± SD AUROC
    # --------------------------------------------------------

    mean_auc = np.mean(
        auc_scores
    )


    if len(auc_scores) > 1:

        std_auc = np.std(
            auc_scores,
            ddof=1
        )

    else:

        std_auc = 0.0


    # --------------------------------------------------------
    # Plot mean ROC
    # --------------------------------------------------------

    line, = plt.plot(

        mean_fpr,

        mean_tpr,

        linewidth=2,

        label=(
            f"{model_name} "
            f"(AUC = "
            f"{mean_auc:.3f} ± "
            f"{std_auc:.3f})"
        )
    )


    # --------------------------------------------------------
    # ±1 SD shaded area
    # --------------------------------------------------------

    plt.fill_between(

        mean_fpr,

        lower_tpr,

        upper_tpr,

        alpha=0.12,

        color=line.get_color()
    )


# Chance line

plt.plot(

    [0, 1],

    [0, 1],

    linestyle="--",

    linewidth=1.5,

    label="Chance (AUC = 0.500)"
)


plt.xlabel(
    "False Positive Rate",
    fontsize=13
)


plt.ylabel(
    "True Positive Rate",
    fontsize=13
)


plt.title(

    "Mean ROC ± 1 SD Across Rolling Windows\n"
    f"{TRAIN_DAYS}-Day Train + "
    f"{TEST_DAYS}-Day Test",

    fontsize=14
)


plt.xlim(0, 1)

plt.ylim(0, 1.02)


plt.legend(
    loc="lower right",
    fontsize=10
)


plt.grid(
    alpha=0.2
)


plt.tight_layout()


plt.savefig(

    "roc_mean_std_rolling_windows.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# ============================================================
# 22. PRINT AUROC OF EVERY ROLLING WINDOW
# ============================================================

print("\n")
print("=" * 100)
print("AUROC ACROSS INDIVIDUAL ROLLING WINDOWS")
print("=" * 100)


for model_name in model_order:

    roc_data = roc_cv_results[
        model_name
    ]


    auc_scores = roc_data[
        "auc_scores"
    ]


    fold_numbers = roc_data[
        "fold_numbers"
    ]


    if len(auc_scores) == 0:
        continue


    print(
        f"\n{model_name}"
    )


    for fold, auc_value in zip(
        fold_numbers,
        auc_scores
    ):

        print(
            f"  Fold {fold:2d}: "
            f"AUROC = {auc_value:.4f}"
        )


    if len(auc_scores) > 1:

        auc_std = np.std(
            auc_scores,
            ddof=1
        )

    else:

        auc_std = 0.0


    print(
        f"  Mean ± SD: "
        f"{np.mean(auc_scores):.4f} "
        f"± {auc_std:.4f}"
    )


# ============================================================
# 23. SHOW TEMPORAL PERFORMANCE OF EACH MODEL
# ============================================================

plt.figure(
    figsize=(10, 6)
)


for model_name in model_order:

    if model_name == "Majority Dummy":
        continue


    model_df = fold_results_df[
        fold_results_df["Model"]
        == model_name
    ]


    # Only windows with valid AUROC
    model_df = model_df.dropna(
        subset=["AUROC"]
    )


    plt.plot(

        model_df["Fold"],

        model_df["AUROC"],

        marker="o",

        label=model_name
    )


plt.axhline(
    0.5,
    linestyle="--",
    linewidth=1
)


plt.xlabel(
    "Rolling Test Window",
    fontsize=12
)


plt.ylabel(
    "AUROC",
    fontsize=12
)


plt.title(
    "AUROC Across Time",
    fontsize=14
)


plt.ylim(
    0,
    1.02
)


plt.legend()


plt.grid(
    alpha=0.2
)


plt.tight_layout()


plt.savefig(

    "auroc_over_time.png",

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print("\n")
print("=" * 90)
print("ANALYSIS COMPLETE")
print("=" * 90)

print(
    "Number of rolling windows:",
    len(time_splits)
)

print(
    f"Training history: {TRAIN_DAYS} days"
)

print(
    f"Testing horizon : {TEST_DAYS} days"
)

print(
    f"Window step     : {STEP_DAYS} days"
)
```
